### Order table cleaning and featuring engineering

```python
orders = pd.read_csv('orders.csv')
orders.head(2)
```

```python
# total rows and columns
orders.shape
```

```python
orders.duplicated().sum()
```

```python
# to check null values and datatypes
# here we need to change datatype of date and time columns
orders.info()
orders.isna().sum()
```

```python
# to combine date and time columns in new column with proper datatype
orders['datetime'] = pd.to_datetime(orders['date'] + ' ' + orders['time'], format='%d-%m-%Y %H:%M:%S')
```

```python
orders.head(2)
```

```python
# to change the datatype of date column
orders['date'] = pd.to_datetime(orders['date'], format='%d-%m-%Y')
```

```python
orders.head(2)
```

```python
# now here I am droping the time column because dtype of time column is object
orders = orders.drop('time', axis=1)
```

```python
#  now i am extracting time column from datetime column
orders['time'] = orders['datetime'].dt.time
orders.head(2)
```

```python
# extracting all datetime features
orders['hour'] = orders['datetime'].dt.hour
orders['day_name'] = orders['datetime'].dt.day_name()
orders['month_name'] = orders['datetime'].dt.month_name()
orders['day'] = orders['datetime'].dt.dayofweek + 1
orders['month'] = orders['datetime'].dt.month
orders['year'] = orders['datetime'].dt.year
orders.head(2)
```

```python
# reordering columns name
# orders = orders[['order_id', 'date', 'time', 'datetime', 'hour', 'day', 'day_name', 'month', 'month_name', 'year']]
# orders.head(2)
```

```python
orders.info()
```

```python
# save cleaned file
orders.to_csv('clean_orders.csv', index=False)
```

### Cleaning pizza_types table

```python
pizza_types = pd.read_csv('pizza_types.csv', encoding='latin1')
pizza_types.head(2)
```

```python
# total rows and columns
pizza_types.shape
```

```python
pizza_types.duplicated().sum()
```

```python
# to check null values and datatypes
pizza_types.info()
```

```python
#  to remove all whitespaces
for col in pizza_types.select_dtypes(include='object').columns:
    pizza_types[col] = pizza_types[col].str.strip()
```

```python
# to check category
pizza_types['category'].unique()
```

```python
# to convert all ingrediients column in lowercase
pizza_types['ingredients'] = pizza_types['ingredients'].str.lower()
```

```python
pizza_types.head(2)
```

```python
# save cleaned file
pizza_types.to_csv('clean_pizza_types.csv', index=False)
```

### Cleaning pizzas table

```python
pizzas = pd.read_csv('pizzas.csv')
pizzas.head(2)
```

```python
# total rows and columns
pizzas.shape
```

```python
pizzas.duplicated().sum()
```

```python
# to check if prize column contains 0 or -ve values
pizzas[pizzas['price'] <= 0]
```

```python
pizzas.duplicated().sum()
```

```python
# to check null values and data types
pizzas.info()
```

```python
pizzas['size'].unique()
```

```python
# convert datatype of size column into category to define an ordered category to ensure meaningful sorting and correct visualization. 
pizzas['size'] = pizzas['size'].astype(pd.CategoricalDtype(categories=['S', 'M', 'L', 'XL', 'XXL'], ordered=True))
```

```python
pizzas.info()
```

```python
# save cleaned file
pizzas.to_csv('clean_pizzas.csv', index=False)
```

## Merge all the tables

```python
pizza_sales = (
    order_details
    .merge(orders, on="order_id", how="inner")
    .merge(pizzas, on="pizza_id", how="inner")
    .merge(pizza_types, on="pizza_type_id", how="inner")
)
```

```python
pizza_sales.head(2)
```

```python
pizza_sales.isna().sum()
```

```python
pizza_sales.info()
```

```python
# check all column_names
pizza_sales.columns
```

```python
# save file
pizza_sales.to_csv('pizza_sales.csv', index=False)
```

```python
# rename name column to pizza_name
pizza_sales.rename(columns={"name": "pizza_name","day": "day_order"}, inplace=True)
```

```python
# create revenue column
pizza_sales['revenue'] = pizza_sales['price'] * pizza_sales['quantity']
```

```python
# choosing specific columns which is important for dashboard
pizza_sales = pizza_sales[["order_id","datetime","hour","day_order","day_name","month","month_name","year","pizza_name","category","size","price","quantity","revenue"]]
```

```python
pizza_sales.head(3)
```

```python
# save file
pizza_sales.to_csv('clean_pizza_sales.csv', index=False)
```
